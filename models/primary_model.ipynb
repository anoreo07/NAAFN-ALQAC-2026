# ALQAC — Final Outcome Pipeline v2 dùng toàn bộ case evidence

Pipeline không dùng public data:

```text
toàn bộ all_unique_chunks
        ↓
Qwen3.5-9B grounded claim–evidence graph
        ↓
Qwen3-Reranker-8B chọn luật riêng cho từng claim
        ↓
3 Qwen3.5-9B judge passes
        ↓
deterministic claim-to-label voting
        ↓
Aya Expanse 8B verifier cho case bất đồng
        ↓
submission.json
```

Các guardrail quan trọng:

- không fallback sang `PARTIAL_A_WIN`;
- JSON lỗi được retry, vẫn lỗi thì pipeline dừng;
- defense gắn vào claim, không tính như counterclaim;
- mọi quote được kiểm tra có thật trong chunk;
- mọi `chunk_id` và `aid` được kiểm tra;
- `case_evidence` cuối chứa toàn bộ unique chunk IDs đã retrieve.

````python
%%writefile /content/alqac_final_outcome_all_evidence_v2.py

from __future__ import annotations

import gc
import json
import math
import os
import re
import time
import unicodedata
from collections import Counter, defaultdict
from dataclasses import asdict, dataclass, field
from pathlib import Path
from typing import Any, Iterable, Optional, Sequence

import numpy as np
import torch
from transformers import (
    AutoModelForCausalLM,
    AutoModelForMultimodalLM,
    AutoProcessor,
    AutoTokenizer,
    BitsAndBytesConfig,
)


# =============================================================================
# Labels and corpus metadata
# =============================================================================

LABELS = (
    "A_WIN",
    "PARTIAL_A_WIN",
    "PARTIAL_B_WIN",
    "B_WIN",
)

LABEL_INDEX = {label: index for index, label in enumerate(LABELS)}

LABEL_DEFINITIONS = {
    "A_WIN": (
        "Nguyên đơn A được chấp nhận về cơ bản toàn bộ hoặc phần cốt lõi của "
        "yêu cầu chính; không có phản tố cốt lõi của B được chấp nhận."
    ),
    "PARTIAL_A_WIN": (
        "A nhận cứu trợ chính hoặc phần lợi ích có trọng lượng lớn hơn, nhưng "
        "một phần đáng kể yêu cầu bị giảm hoặc bác."
    ),
    "PARTIAL_B_WIN": (
        "A vẫn nhận một cứu trợ có ý nghĩa, nhưng B bác được phần lớn yêu cầu "
        "hoặc thắng vấn đề/phản tố trung tâm."
    ),
    "B_WIN": (
        "Yêu cầu cốt lõi của A bị bác về cơ bản/toàn bộ và A không nhận cứu trợ "
        "thực chất, hoặc phản tố cốt lõi của B thắng rõ rệt."
    ),
}

LAW_TITLES: dict[str, str] = {
    "47/2010/QH12": "Luật Các tổ chức tín dụng 2010",
    "26/2008/QH12": "Luật Thi hành án dân sự 2008",
    "36/2005/QH11": "Luật Thương mại 2005",
    "43/2014/NĐ-CP": "Nghị định 43/2014/NĐ-CP hướng dẫn Luật Đất đai",
    "326/2016/UBTVQH14": "Nghị quyết 326/2016/UBTVQH14 về án phí và lệ phí Tòa án",
    "11/2017/QH14": "Luật Trợ giúp pháp lý 2017",
    "92/2015/QH13": "Bộ luật Tố tụng dân sự 2015",
    "39/2016/TT-NHNN": "Thông tư 39/2016/TT-NHNN về hoạt động cho vay",
    "91/2015/QH13": "Bộ luật Dân sự 2015",
    "52/2014/QH13": "Luật Hôn nhân và gia đình 2014",
    "59/2020/QH14": "Luật Doanh nghiệp 2020",
    "53/2014/QH13": "Luật Phá sản 2014",
    "45/2013/QH13": "Luật Đất đai 2013",
    "46/2010/QH12": "Luật Ngân hàng Nhà nước Việt Nam 2010",
}

PREDICTION_MARKERS = (
    "agent cần dự đoán",
    "agent dự đoán",
    "hãy dự đoán",
    "cần dự đoán",
    "dự đoán ai",
    "ai được chấp nhận yêu cầu",
    "ai thắng kiện",
    "nguyên đơn thắng kiện hay bị đơn",
    "nguyên đơn hay bị đơn thắng",
)

IMPORTANCE_WEIGHT = {
    "CORE": 5.0,
    "MAJOR": 2.5,
    "SECONDARY": 1.0,
}

RESULT_FRACTION = {
    "GRANTED": 1.0,
    "PARTIALLY_GRANTED": 0.55,
    "REJECTED": 0.0,
    "UNCERTAIN": 0.5,
}

VALID_IMPORTANCE = set(IMPORTANCE_WEIGHT)
VALID_RESULTS = set(RESULT_FRACTION)
VALID_CLAIM_OWNERS = {"A", "B"}
VALID_CLAIM_TYPES = {
    "A_CLAIM",
    "B_COUNTERCLAIM",
}
VALID_DEFENSE_EFFECTS = {
    "DEFEAT_CLAIM",
    "REDUCE_RELIEF",
    "NO_EFFECT",
    "UNCERTAIN",
}
VALID_PROCEDURAL_STAGE = {
    "FIRST_INSTANCE",
    "APPEAL",
    "CASSATION_REOPENING",
    "UNKNOWN",
}


# =============================================================================
# Configuration and data classes
# =============================================================================

@dataclass
class FinalOutcomeConfig:
    qwen_model_id: str = "Qwen/Qwen3.5-9B"
    reranker_model_id: str = "Qwen/Qwen3-Reranker-8B"
    verifier_model_id: str = "CohereLabs/aya-expanse-8b"
    verifier_fallback_model_id: str = "Qwen/Qwen3-8B"

    qwen_quant_mode: str = "bf16"  # bf16 | bnb4 | bnb8
    reranker_quant_mode: str = "bf16"
    verifier_quant_mode: str = "bf16"

    extraction_enable_thinking: bool = False
    judge_enable_thinking: bool = True
    deterministic_generation: bool = True

    max_json_retries: int = 3
    extraction_max_new_tokens: int = 1700
    coverage_audit_max_new_tokens: int = 900
    judge_max_new_tokens: int = 1250
    verifier_max_new_tokens: int = 850

    # Every unique case chunk is used in extraction and is submitted.
    use_all_unique_chunks: bool = True
    run_conditional_coverage_audit: bool = True
    coverage_audit_min_claims: int = 1
    coverage_audit_min_grounded_quotes: int = 4

    max_laws_per_claim: int = 4
    reranker_batch_size: int = 4
    reranker_max_length: int = 4096

    # Three complementary adjudication passes.
    judge_passes: tuple[str, ...] = (
        "ELEMENTS_BURDEN",
        "ADVERSARIAL",
        "REMEDY_OUTCOME",
    )

    # Verifier runs only when consensus is not strong.
    verifier_vote_margin_threshold: int = 3
    verifier_rule_disagreement: bool = True
    allow_verifier_fallback: bool = True
    hf_token: Optional[str] = None

    # Deterministic mapping.
    a_win_core_threshold: float = 0.85
    b_win_core_threshold: float = 0.15
    partial_a_share_threshold: float = 0.56
    partial_b_share_threshold: float = 0.44

    expected_private_cases: int = 60


@dataclass
class LawArticle:
    aid: int
    law_id: str
    law_title: str
    article_number: int
    text: str


@dataclass
class CaseBundle:
    case_id: str
    case_query: str
    clean_query: str
    chunks: list[dict[str, Any]]
    all_chunk_ids: list[str]
    selected_law_aids: list[int]
    hints: dict[str, Any]

    full_evidence_text: str = ""
    graph: dict[str, Any] = field(default_factory=dict)
    claim_laws: dict[str, list[dict[str, Any]]] = field(default_factory=dict)


# =============================================================================
# Generic helpers
# =============================================================================

def normalize_text(value: Any) -> str:
    text = unicodedata.normalize("NFC", str(value or "")).lower()
    text = re.sub(r"\s+", " ", text)
    return text.strip()


def normalize_for_quote(value: Any) -> str:
    text = unicodedata.normalize("NFC", str(value or "")).lower()
    text = re.sub(r"[^\w\s]", " ", text, flags=re.UNICODE)
    text = re.sub(r"\s+", " ", text)
    return text.strip()


def clean_case_query(case_query: str) -> str:
    text = " ".join(str(case_query or "").split())
    lowered = text.lower()
    positions = [
        lowered.find(marker)
        for marker in PREDICTION_MARKERS
        if lowered.find(marker) >= 0
    ]
    if positions:
        text = text[:min(positions)]
    return text.strip(" .?;:")


def truncate_words(value: Any, max_words: int) -> str:
    words = str(value or "").split()
    if len(words) <= max_words:
        return " ".join(words)
    return " ".join(words[:max_words]) + " …"


def dedupe_preserve(values: Iterable[Any]) -> list[Any]:
    output = []
    seen = set()
    for value in values:
        marker = (
            json.dumps(value, ensure_ascii=False, sort_keys=True)
            if isinstance(value, (dict, list))
            else value
        )
        if marker in seen:
            continue
        seen.add(marker)
        output.append(value)
    return output


def atomic_write_json(path: Path, value: Any) -> None:
    path.parent.mkdir(parents=True, exist_ok=True)
    temp = path.with_suffix(path.suffix + ".tmp")
    temp.write_text(
        json.dumps(value, ensure_ascii=False, indent=2),
        encoding="utf-8",
    )
    temp.replace(path)


def read_json(path: Path) -> Any:
    return json.loads(path.read_text(encoding="utf-8-sig"))


def numeric_case_key(case_id: str) -> tuple[int, str]:
    match = re.search(r"\d+", str(case_id))
    return (
        int(match.group()) if match else 10**9,
        str(case_id),
    )


def safe_float(value: Any, default: float = 0.0) -> float:
    try:
        result = float(value)
    except (TypeError, ValueError):
        return default
    if not math.isfinite(result):
        return default
    return result


def clamp(value: float, minimum: float, maximum: float) -> float:
    return max(minimum, min(maximum, value))


def probability_margin(probabilities: dict[str, float]) -> float:
    values = sorted(probabilities.values(), reverse=True)
    if len(values) < 2:
        return 1.0
    return float(values[0] - values[1])


def top_label(probabilities: dict[str, float]) -> str:
    return max(
        LABELS,
        key=lambda label: (
            probabilities.get(label, 0.0),
            -LABEL_INDEX[label],
        ),
    )


# =============================================================================
# Corpus and private files
# =============================================================================

def load_law_corpus(path: Path) -> dict[int, LawArticle]:
    payload = read_json(path)
    output: dict[int, LawArticle] = {}

    for document in payload:
        law_id = str(document["law_id"])
        law_title = LAW_TITLES.get(law_id, law_id)
        for article_number, article in enumerate(
            document.get("content", []),
            start=1,
        ):
            aid = int(article["aid"])
            output[aid] = LawArticle(
                aid=aid,
                law_id=law_id,
                law_title=law_title,
                article_number=article_number,
                text=str(
                    article.get("content_Article", "")
                ).strip(),
            )

    if not output:
        raise ValueError("Corpus luật rỗng.")
    return output


def find_case_files(
    directory: Path,
    required_fragment: str,
) -> dict[str, Path]:
    output: dict[str, Path] = {}
    for path in directory.rglob("*.json"):
        lower = path.name.lower()
        if required_fragment not in lower:
            continue
        match = re.search(r"(case_\d+)", lower)
        if not match:
            continue
        case_id = match.group(1)
        current = output.get(case_id)
        if current is None or len(str(path)) < len(str(current)):
            output[case_id] = path
    return output


def load_law_final_rows(path: Path) -> dict[str, dict[str, Any]]:
    payload = read_json(path)
    if not isinstance(payload, list):
        raise ValueError(
            "private_case_and_law_evidence.json phải là JSON array."
        )

    output = {}
    for row in payload:
        case_id = str(row.get("case_id", ""))
        if not case_id:
            continue
        if case_id in output:
            raise ValueError(f"Trùng case_id trong law output: {case_id}")
        output[case_id] = row
    return output


def load_private_cases(
    private_test_path: Path,
    case_retrieval_dir: Path,
    private_case_and_law_path: Path,
) -> tuple[list[CaseBundle], list[dict[str, Any]]]:
    originals = read_json(private_test_path)
    evidence_files = find_case_files(
        case_retrieval_dir,
        "evidence_result",
    )
    handoff_files = find_case_files(
        case_retrieval_dir,
        "law_handoff",
    )
    law_rows = load_law_final_rows(private_case_and_law_path)

    cases = []
    for original in originals:
        case_id = str(original["case_id"])
        evidence_path = evidence_files.get(case_id)
        if evidence_path is None:
            raise FileNotFoundError(
                f"Thiếu evidence_result cho {case_id}"
            )

        result = read_json(evidence_path)
        handoff = (
            read_json(handoff_files[case_id])
            if case_id in handoff_files
            else result.get("handoff_for_later_stages", {})
        )
        law_row = law_rows.get(case_id)
        if law_row is None:
            raise ValueError(
                f"Thiếu law retrieval output cho {case_id}"
            )

        chunks = result.get("all_unique_chunks", [])
        if not isinstance(chunks, list):
            chunks = []

        cleaned_chunks = []
        seen_ids = set()
        for position, chunk in enumerate(chunks, start=1):
            chunk_id = str(chunk.get("chunk_id", "")).strip()
            text = " ".join(str(chunk.get("text", "")).split())
            if not chunk_id or not text or chunk_id in seen_ids:
                continue
            seen_ids.add(chunk_id)
            cleaned_chunks.append({
                **chunk,
                "chunk_id": chunk_id,
                "text": text,
                "_retrieval_position": position,
            })

        law_aids = []
        for value in law_row.get("law_evidence", []):
            if isinstance(value, dict):
                value = value.get("aid")
            try:
                law_aids.append(int(value))
            except (TypeError, ValueError):
                continue

        hints = {
            "atomic_facts": handoff.get(
                "atomic_facts",
                result.get("atomic_facts", []),
            ),
            "evidence_topics": handoff.get(
                "evidence_topics",
                result.get("evidence_topics", []),
            ),
            "unresolved_factual_gaps": handoff.get(
                "unresolved_factual_gaps",
                result.get("unresolved_factual_gaps", []),
            ),
            "source_coverage": handoff.get(
                "source_coverage",
                result.get("source_coverage", {}),
            ),
        }

        cases.append(CaseBundle(
            case_id=case_id,
            case_query=str(
                result.get(
                    "case_query",
                    original.get("case_query", ""),
                )
            ),
            clean_query=str(
                result.get(
                    "clean_query",
                    clean_case_query(
                        original.get("case_query", "")
                    ),
                )
            ),
            chunks=cleaned_chunks,
            all_chunk_ids=[
                chunk["chunk_id"]
                for chunk in cleaned_chunks
            ],
            selected_law_aids=dedupe_preserve(law_aids),
            hints=hints,
        ))

    return cases, originals


# =============================================================================
# Full-evidence packet
# =============================================================================

def observed_source_label(chunk: dict[str, Any]) -> str:
    observed = chunk.get("observed_source_types", [])
    if isinstance(observed, list) and observed:
        return "/".join(str(value) for value in observed)
    return "unknown"


def build_full_evidence_text(case: CaseBundle) -> str:
    blocks = []
    for index, chunk in enumerate(case.chunks, start=1):
        blocks.append(
            f"[EVIDENCE {index}; chunk_id={chunk['chunk_id']}; "
            f"observed_source={observed_source_label(chunk)}]\n"
            f"{chunk['text']}"
        )
    case.full_evidence_text = "\n\n".join(blocks)
    return case.full_evidence_text


def build_hint_text(hints: dict[str, Any]) -> str:
    parts = []

    facts = hints.get("atomic_facts", [])
    if isinstance(facts, list):
        rows = []
        for item in facts[:14]:
            value = (
                item.get("fact", "")
                if isinstance(item, dict)
                else item
            )
            value = " ".join(str(value).split())
            if value:
                rows.append(f"- {value}")
        if rows:
            parts.append(
                "FACT HINTS DO PIPELINE TRƯỚC SINH "
                "(không mặc định đúng):\n"
                + "\n".join(rows)
            )

    topics = hints.get("evidence_topics", [])
    if isinstance(topics, list):
        rows = []
        for item in topics[:12]:
            value = (
                item.get("title")
                or item.get("description")
                or ""
            ) if isinstance(item, dict) else item
            value = " ".join(str(value).split())
            if value:
                rows.append(f"- {value}")
        if rows:
            parts.append(
                "TOPIC HINTS:\n" + "\n".join(rows)
            )

    gaps = hints.get("unresolved_factual_gaps", [])
    if isinstance(gaps, list):
        rows = []
        for item in gaps[:12]:
            value = (
                item.get("description")
                or item.get("gap")
                or ""
            ) if isinstance(item, dict) else item
            value = " ".join(str(value).split())
            if value:
                rows.append(f"- {value}")
        if rows:
            parts.append(
                "ĐIỂM CHƯA CHẮC CHẮN:\n"
                + "\n".join(rows)
            )

    return "\n\n".join(parts)


# =============================================================================
# Runtime helpers
# =============================================================================

def quantization_kwargs(mode: str) -> dict[str, Any]:
    normalized = mode.lower()
    if normalized == "bf16":
        return {"torch_dtype": torch.bfloat16}
    if normalized == "bnb4":
        return {
            "quantization_config": BitsAndBytesConfig(
                load_in_4bit=True,
                bnb_4bit_compute_dtype=torch.bfloat16,
                bnb_4bit_quant_type="nf4",
                bnb_4bit_use_double_quant=False,
            ),
        }
    if normalized == "bnb8":
        return {
            "quantization_config": BitsAndBytesConfig(
                load_in_8bit=True,
            ),
        }
    raise ValueError(
        "Quant mode phải là bf16, bnb4 hoặc bnb8."
    )


class JsonMixin:
    @staticmethod
    def extract_json_object(text: str) -> dict[str, Any]:
        cleaned = str(text or "").strip()
        cleaned = re.sub(
            r"^```(?:json)?\s*",
            "",
            cleaned,
            flags=re.IGNORECASE,
        )
        cleaned = re.sub(r"\s*```$", "", cleaned)

        decoder = json.JSONDecoder()
        for index, char in enumerate(cleaned):
            if char != "{":
                continue
            try:
                value, _ = decoder.raw_decode(cleaned[index:])
                if isinstance(value, dict):
                    return value
            except json.JSONDecodeError:
                continue

        raise ValueError(
            "Không parse được JSON:\n" + cleaned[:2200]
        )


class QwenRuntime(JsonMixin):
    def __init__(self, config: FinalOutcomeConfig):
        self.config = config
        self.processor: Optional[Any] = None
        self.model: Optional[Any] = None

    def load(self) -> None:
        if not torch.cuda.is_available():
            raise RuntimeError("Hãy chọn A100 GPU.")

        torch.backends.cuda.matmul.allow_tf32 = True
        if hasattr(torch.backends, "cudnn"):
            torch.backends.cudnn.allow_tf32 = True

        self.processor = AutoProcessor.from_pretrained(
            self.config.qwen_model_id,
        )
        kwargs = {
            "device_map": "auto",
            "low_cpu_mem_usage": True,
            **quantization_kwargs(
                self.config.qwen_quant_mode
            ),
        }
        self.model = AutoModelForMultimodalLM.from_pretrained(
            self.config.qwen_model_id,
            **kwargs,
        )
        self.model.eval()
        print("Loaded:", self.config.qwen_model_id)

    def unload(self) -> None:
        self.processor = None
        self.model = None
        gc.collect()
        torch.cuda.empty_cache()

    def generate_json(
        self,
        system: str,
        user: str,
        *,
        max_new_tokens: int,
        enable_thinking: bool,
    ) -> dict[str, Any]:
        if self.model is None or self.processor is None:
            raise RuntimeError("Qwen runtime chưa load.")

        messages = [
            {
                "role": "system",
                "content": [{
                    "type": "text",
                    "text": system,
                }],
            },
            {
                "role": "user",
                "content": [{
                    "type": "text",
                    "text": user,
                }],
            },
        ]
        common = {
            "add_generation_prompt": True,
            "tokenize": True,
            "return_dict": True,
            "return_tensors": "pt",
        }
        try:
            inputs = self.processor.apply_chat_template(
                messages,
                enable_thinking=enable_thinking,
                **common,
            )
        except TypeError:
            inputs = self.processor.apply_chat_template(
                messages,
                chat_template_kwargs={
                    "enable_thinking": enable_thinking,
                },
                **common,
            )

        inputs = inputs.to(self.model.device)
        input_length = inputs["input_ids"].shape[-1]
        kwargs = {
            "max_new_tokens": max_new_tokens,
            "do_sample": not (
                self.config.deterministic_generation
            ),
            "pad_token_id": (
                self.processor.tokenizer.eos_token_id
            ),
            "eos_token_id": (
                self.processor.tokenizer.eos_token_id
            ),
            "use_cache": True,
        }
        if not self.config.deterministic_generation:
            kwargs.update({
                "temperature": 0.25,
                "top_p": 0.85,
            })

        with torch.inference_mode():
            generated = self.model.generate(
                **inputs,
                **kwargs,
            )

        new_tokens = generated[0, input_length:]
        text = self.processor.decode(
            new_tokens,
            skip_special_tokens=True,
        ).strip()
        return self.extract_json_object(text)


class QwenRerankerRuntime:
    """
    Official yes/no-logit scoring pattern for Qwen3-Reranker.
    """

    def __init__(self, config: FinalOutcomeConfig):
        self.config = config
        self.tokenizer: Optional[Any] = None
        self.model: Optional[Any] = None
        self.token_no_id: Optional[int] = None
        self.token_yes_id: Optional[int] = None
        self.prefix_tokens: list[int] = []
        self.suffix_tokens: list[int] = []

    def load(self) -> None:
        self.tokenizer = AutoTokenizer.from_pretrained(
            self.config.reranker_model_id,
            padding_side="left",
        )
        kwargs = {
            "device_map": "auto",
            "low_cpu_mem_usage": True,
            **quantization_kwargs(
                self.config.reranker_quant_mode
            ),
        }
        self.model = AutoModelForCausalLM.from_pretrained(
            self.config.reranker_model_id,
            **kwargs,
        )
        self.model.eval()

        self.token_no_id = self.tokenizer.convert_tokens_to_ids(
            "no"
        )
        self.token_yes_id = self.tokenizer.convert_tokens_to_ids(
            "yes"
        )
        prefix = (
            "<|im_start|>system\n"
            "Judge whether the Document meets the requirements based on "
            "the Query and the Instruct provided. Note that the answer can "
            "only be \"yes\" or \"no\"."
            "<|im_end|>\n<|im_start|>user\n"
        )
        suffix = (
            "<|im_end|>\n<|im_start|>assistant\n"
            "<think>\n\n</think>\n\n"
        )
        self.prefix_tokens = self.tokenizer.encode(
            prefix,
            add_special_tokens=False,
        )
        self.suffix_tokens = self.tokenizer.encode(
            suffix,
            add_special_tokens=False,
        )
        print("Loaded:", self.config.reranker_model_id)

    def unload(self) -> None:
        self.tokenizer = None
        self.model = None
        gc.collect()
        torch.cuda.empty_cache()

    def _format_pair(
        self,
        query: str,
        document: str,
    ) -> str:
        instruction = (
            "Given a Vietnamese civil-case claim and its grounded evidence, "
            "judge whether the statutory provision directly governs the "
            "claim, its legal elements, remedy, defense, burden of proof, "
            "or procedural disposition."
        )
        return (
            f"<Instruct>: {instruction}\n"
            f"<Query>: {query}\n"
            f"<Document>: {document}"
        )

    def score_pairs(
        self,
        pairs: Sequence[tuple[str, str]],
    ) -> list[float]:
        if self.model is None or self.tokenizer is None:
            raise RuntimeError("Reranker chưa load.")

        all_scores: list[float] = []
        batch_size = self.config.reranker_batch_size
        max_length = self.config.reranker_max_length

        for start in range(0, len(pairs), batch_size):
            batch = pairs[start:start + batch_size]
            texts = [
                self._format_pair(query, doc)
                for query, doc in batch
            ]
            inputs = self.tokenizer(
                texts,
                padding=False,
                truncation="longest_first",
                return_attention_mask=False,
                max_length=(
                    max_length
                    - len(self.prefix_tokens)
                    - len(self.suffix_tokens)
                ),
            )
            for index, token_ids in enumerate(
                inputs["input_ids"]
            ):
                inputs["input_ids"][index] = (
                    self.prefix_tokens
                    + token_ids
                    + self.suffix_tokens
                )

            padded = self.tokenizer.pad(
                inputs,
                padding=True,
                return_tensors="pt",
                max_length=max_length,
            )
            padded = {
                key: value.to(self.model.device)
                for key, value in padded.items()
            }

            with torch.inference_mode():
                logits = self.model(**padded).logits[:, -1, :]

            no_vector = logits[:, self.token_no_id]
            yes_vector = logits[:, self.token_yes_id]
            stacked = torch.stack(
                [no_vector, yes_vector],
                dim=1,
            )
            probabilities = torch.softmax(
                stacked,
                dim=1,
            )[:, 1]
            all_scores.extend(
                float(value)
                for value in probabilities.detach().cpu()
            )

        return all_scores


class AyaVerifierRuntime(JsonMixin):
    def __init__(self, config: FinalOutcomeConfig):
        self.config = config
        self.model_id = config.verifier_model_id
        self.tokenizer: Optional[Any] = None
        self.model: Optional[Any] = None
        self.using_fallback = False

    def _load_model(self, model_id: str) -> None:
        token = self.config.hf_token or os.getenv("HF_TOKEN")
        self.tokenizer = AutoTokenizer.from_pretrained(
            model_id,
            token=token,
        )
        kwargs = {
            "device_map": "auto",
            "low_cpu_mem_usage": True,
            "token": token,
            **quantization_kwargs(
                self.config.verifier_quant_mode
            ),
        }
        self.model = AutoModelForCausalLM.from_pretrained(
            model_id,
            **kwargs,
        )
        self.model.eval()
        self.model_id = model_id

    def load(self) -> None:
        try:
            self._load_model(
                self.config.verifier_model_id
            )
        except Exception as exc:
            if not self.config.allow_verifier_fallback:
                raise
            print(
                "Không load được Aya:",
                type(exc).__name__,
                exc,
            )
            print(
                "Dùng fallback verifier:",
                self.config.verifier_fallback_model_id,
            )
            self.using_fallback = True
            self._load_model(
                self.config.verifier_fallback_model_id
            )
        print(
            "Loaded verifier:",
            self.model_id,
            "fallback=",
            self.using_fallback,
        )

    def unload(self) -> None:
        self.tokenizer = None
        self.model = None
        gc.collect()
        torch.cuda.empty_cache()

    def generate_json(
        self,
        system: str,
        user: str,
        max_new_tokens: int,
    ) -> dict[str, Any]:
        if self.model is None or self.tokenizer is None:
            raise RuntimeError("Verifier chưa load.")

        messages = [
            {"role": "system", "content": system},
            {"role": "user", "content": user},
        ]
        common = {
            "tokenize": True,
            "add_generation_prompt": True,
            "return_dict": True,
            "return_tensors": "pt",
        }
        try:
            inputs = self.tokenizer.apply_chat_template(
                messages,
                enable_thinking=False,
                **common,
            )
        except TypeError:
            inputs = self.tokenizer.apply_chat_template(
                messages,
                **common,
            )

        inputs = inputs.to(self.model.device)
        input_length = inputs["input_ids"].shape[-1]

        kwargs = {
            "max_new_tokens": max_new_tokens,
            "do_sample": False,
            "pad_token_id": self.tokenizer.eos_token_id,
            "eos_token_id": self.tokenizer.eos_token_id,
            "use_cache": True,
        }

        with torch.inference_mode():
            generated = self.model.generate(
                **inputs,
                **kwargs,
            )
        new_tokens = generated[0, input_length:]
        text = self.tokenizer.decode(
            new_tokens,
            skip_special_tokens=True,
        ).strip()
        return self.extract_json_object(text)


# =============================================================================
# Grounded evidence graph
# =============================================================================

GRAPH_SYSTEM = """
Bạn là bộ trích xuất claim–evidence graph cho vụ án dân sự Việt Nam.

A luôn là NGUYÊN ĐƠN; B luôn là BỊ ĐƠN, kể cả ở phúc thẩm.

Nhiệm vụ chỉ là tổ chức dữ kiện, KHÔNG dự đoán nhãn thắng thua.

Quy tắc grounding:
1. Đọc toàn bộ raw evidence.
2. Mỗi evidence item phải có chunk_id chính xác và quote nguyên văn từ chunk.
3. Phân biệt lời khai/cáo buộc với thừa nhận và chứng cứ khách quan.
4. B_DEFENSE phải gắn vào A_CLAIM mà nó chống lại; không biến defense thành
   counterclaim độc lập.
5. B_COUNTERCLAIM chỉ tồn tại khi B yêu cầu Tòa trao một cứu trợ độc lập cho B.
6. BẮT BUỘC luôn có ít nhất một claim_type="A_CLAIM". CASE QUERY đã mô tả
   yêu cầu của nguyên đơn; kể cả khi raw chunks thiếu đoạn lời khai nguyên đơn,
   hãy dùng yêu cầu trong CASE QUERY để tạo A_CLAIM. Evidence arrays có thể rỗng
   nếu chưa tìm được exact quote, nhưng không được bỏ A_CLAIM.
7. Chấp nhận nhiều cách gọi vai trong evidence, nhưng output owner chỉ được là
   "A" hoặc "B", claim_type chỉ được là "A_CLAIM" hoặc
   "B_COUNTERCLAIM".
8. Phân biệt quyết định cấp dưới, nội dung kháng cáo, ý kiến Viện kiểm sát và
   kết quả cuối cần dự đoán.
9. Facts/topics/gaps do pipeline trước sinh chỉ là hints.
10. Không thêm điều luật hoặc facts không có trong input.
11. Chỉ trả JSON hợp lệ.
"""


def graph_schema_text() -> str:
    return """
Schema:
{
  "procedural_stage": "FIRST_INSTANCE|APPEAL|CASSATION_REOPENING|UNKNOWN",
  "appeal_scope": [
    {
      "description": "...",
      "chunk_id": "...",
      "quote": "..."
    }
  ],
  "prior_decisions": [
    {
      "court_level": "...",
      "decision": "...",
      "chunk_id": "...",
      "quote": "..."
    }
  ],
  "prosecutor_opinions": [
    {
      "opinion": "...",
      "chunk_id": "...",
      "quote": "..."
    }
  ],
  "claims": [
    {
      "claim_id": "A1 hoặc B1",
      "owner": "A|B",
      "claim_type": "A_CLAIM|B_COUNTERCLAIM",
      "claim": "...",
      "remedy": "...",
      "importance": "CORE|MAJOR|SECONDARY",
      "amount_or_property": "...",
      "supporting_evidence": [
        {
          "chunk_id": "...",
          "quote": "...",
          "evidence_status": "ALLEGED|ADMITTED|CORROBORATED|OBJECTIVE"
        }
      ],
      "opposing_evidence": [
        {
          "chunk_id": "...",
          "quote": "...",
          "evidence_status": "DENIAL|CONTRADICTION|OBJECTIVE"
        }
      ],
      "defenses": [
        {
          "defense": "...",
          "effect_candidate": "DEFEAT_CLAIM|REDUCE_RELIEF|NO_EFFECT|UNCERTAIN",
          "evidence": [
            {"chunk_id": "...", "quote": "..."}
          ]
        }
      ],
      "uncertainties": ["..."]
    }
  ],
  "admissions": [
    {"party": "A|B", "fact": "...", "chunk_id": "...", "quote": "..."}
  ],
  "objective_facts": [
    {"fact": "...", "chunk_id": "...", "quote": "..."}
  ],
  "coverage_notes": ["..."]
}
"""


def quote_is_grounded(
    chunk_id: str,
    quote: str,
    chunk_by_id: dict[str, dict[str, Any]],
) -> bool:
    if chunk_id not in chunk_by_id:
        return False
    normalized_quote = normalize_for_quote(quote)
    if len(normalized_quote.split()) < 3:
        return False
    normalized_chunk = normalize_for_quote(
        chunk_by_id[chunk_id]["text"]
    )
    return normalized_quote in normalized_chunk


def validate_quote_item(
    item: Any,
    chunk_by_id: dict[str, dict[str, Any]],
) -> Optional[dict[str, Any]]:
    if not isinstance(item, dict):
        return None
    chunk_id = str(item.get("chunk_id", "")).strip()
    quote = " ".join(str(item.get("quote", "")).split())
    if not quote_is_grounded(
        chunk_id,
        quote,
        chunk_by_id,
    ):
        return None
    return {
        **item,
        "chunk_id": chunk_id,
        "quote": quote,
    }



def normalize_graph_owner(value: Any) -> str:
    """
    Normalize common model outputs to the ALQAC A/B convention.

    Qwen may return PLAINTIFF, NGUYÊN ĐƠN, claimant names, or lowercase values
    even when the schema asks for A/B. The old validator silently dropped such
    claims, which could remove every plaintiff claim.
    """
    normalized = normalize_for_quote(value)
    compact = normalized.replace(" ", "_")

    a_aliases = {
        "a",
        "plaintiff",
        "claimant",
        "nguyen_don",
        "nguoi_khoi_kien",
        "ben_khoi_kien",
    }
    b_aliases = {
        "b",
        "defendant",
        "respondent",
        "bi_don",
        "nguoi_bi_kien",
        "ben_bi_kien",
    }

    if compact in a_aliases:
        return "A"
    if compact in b_aliases:
        return "B"

    if (
        "nguyen don" in normalized
        or "plaintiff" in normalized
        or "claimant" in normalized
    ):
        return "A"
    if (
        "bi don" in normalized
        or "defendant" in normalized
        or "respondent" in normalized
    ):
        return "B"
    return ""


def normalize_graph_claim_type(
    value: Any,
    owner: str,
) -> str:
    normalized = normalize_for_quote(value)
    compact = normalized.replace(" ", "_")

    if compact in {
        "a_claim",
        "plaintiff_claim",
        "claimant_claim",
        "claim_of_a",
        "yeu_cau_nguyen_don",
        "yeu_cau_cua_nguyen_don",
    }:
        return "A_CLAIM"

    if compact in {
        "b_counterclaim",
        "defendant_counterclaim",
        "respondent_counterclaim",
        "counterclaim",
        "counter_claim",
        "phan_to",
        "yeu_cau_phan_to",
    }:
        return "B_COUNTERCLAIM"

    if (
        "counterclaim" in normalized
        or "phan to" in normalized
    ):
        return "B_COUNTERCLAIM"

    if (
        "plaintiff" in normalized
        or "nguyen don" in normalized
        or owner == "A"
    ):
        return "A_CLAIM"

    if owner == "B":
        # Do not turn a mere defense into a counterclaim. Only accept B's row
        # as a counterclaim when the model explicitly indicates an independent
        # request for relief.
        if any(
            marker in normalized
            for marker in (
                "counter",
                "phan to",
                "independent relief",
                "yeu cau doc lap",
            )
        ):
            return "B_COUNTERCLAIM"
        return ""

    return ""


def extract_query_plaintiff_claim(
    case: CaseBundle,
) -> tuple[str, str]:
    """
    Create a mandatory A-claim seed from the trusted private-test query.

    The query explicitly states what the plaintiff requests. This seed is not
    presented as proven evidence; its supporting_evidence remains empty until
    grounded raw chunks are found.
    """
    query = " ".join(
        str(case.clean_query or case.case_query).split()
    )

    patterns = (
        r"(?:nguyên đơn|các nguyên đơn)\s+yêu cầu\s+(.+)$",
        r"(?:ông|bà|công ty|ngân hàng|quỹ)\s+[^.;]{0,90}\s+yêu cầu\s+(.+)$",
        r"(?:khởi kiện|đề nghị)\s+(.+)$",
    )

    request = ""
    for pattern in patterns:
        match = re.search(
            pattern,
            query,
            flags=re.IGNORECASE,
        )
        if match:
            request = match.group(1).strip(" .;:")
            break

    if not request:
        request = query

    claim_text = truncate_words(request, 70)
    remedy = truncate_words(request, 55)
    return claim_text, remedy


def build_query_seed_a_claim(
    case: CaseBundle,
    existing_claim_ids: set[str],
) -> dict[str, Any]:
    claim_id = "A_QUERY_SEED"
    suffix = 1
    while claim_id in existing_claim_ids:
        suffix += 1
        claim_id = f"A_QUERY_SEED_{suffix}"

    claim_text, remedy = extract_query_plaintiff_claim(case)
    return {
        "claim_id": claim_id,
        "owner": "A",
        "claim_type": "A_CLAIM",
        "claim": claim_text,
        "remedy": remedy,
        "importance": "CORE",
        "amount_or_property": "",
        "supporting_evidence": [],
        "opposing_evidence": [],
        "defenses": [],
        "uncertainties": [
            "Claim được seed từ case_query; cần đánh giá bằng toàn bộ raw evidence."
        ],
        "_seeded_from_case_query": True,
    }


def validate_graph(
    raw: dict[str, Any],
    case: CaseBundle,
    *,
    require_a_claim: bool = True,
) -> dict[str, Any]:
    chunk_by_id = {
        chunk["chunk_id"]: chunk
        for chunk in case.chunks
    }

    stage = str(
        raw.get("procedural_stage", "UNKNOWN")
    ).upper()
    if stage not in VALID_PROCEDURAL_STAGE:
        stage = "UNKNOWN"

    graph: dict[str, Any] = {
        "case_id": case.case_id,
        "procedural_stage": stage,
        "appeal_scope": [],
        "prior_decisions": [],
        "prosecutor_opinions": [],
        "claims": [],
        "admissions": [],
        "objective_facts": [],
        "coverage_notes": [
            str(value)
            for value in raw.get("coverage_notes", [])
            if str(value).strip()
        ][:20],
    }

    for field_name in (
        "appeal_scope",
        "prior_decisions",
        "prosecutor_opinions",
        "admissions",
        "objective_facts",
    ):
        values = raw.get(field_name, [])
        if not isinstance(values, list):
            continue
        for item in values:
            valid = validate_quote_item(
                item,
                chunk_by_id,
            )
            if valid is not None:
                graph[field_name].append(valid)

    claims = raw.get("claims", [])
    if not isinstance(claims, list):
        claims = []

    seen_claim_ids = set()
    for index, item in enumerate(claims, start=1):
        if not isinstance(item, dict):
            continue

        owner = normalize_graph_owner(
            item.get("owner", "")
        )
        claim_type = normalize_graph_claim_type(
            item.get("claim_type", ""),
            owner,
        )

        # Explicit claim_type takes precedence and repairs missing owner.
        if claim_type == "A_CLAIM":
            owner = "A"
        elif claim_type == "B_COUNTERCLAIM":
            owner = "B"
        else:
            # Unknown rows are usually defenses, observations, or malformed
            # outputs. They must not be promoted to independent claims.
            continue

        claim_id = str(
            item.get("claim_id", "")
        ).strip() or f"{owner}{index}"
        if claim_id in seen_claim_ids:
            claim_id = f"{claim_id}_{index}"
        seen_claim_ids.add(claim_id)

        importance = str(
            item.get("importance", "SECONDARY")
        ).upper()
        if importance not in VALID_IMPORTANCE:
            importance = "SECONDARY"

        claim = {
            "claim_id": claim_id,
            "owner": owner,
            "claim_type": claim_type,
            "claim": truncate_words(
                item.get("claim", ""),
                70,
            ),
            "remedy": truncate_words(
                item.get("remedy", ""),
                60,
            ),
            "importance": importance,
            "amount_or_property": truncate_words(
                item.get("amount_or_property", ""),
                45,
            ),
            "supporting_evidence": [],
            "opposing_evidence": [],
            "defenses": [],
            "uncertainties": [
                truncate_words(value, 45)
                for value in item.get(
                    "uncertainties",
                    [],
                )
                if str(value).strip()
            ][:12],
        }

        for field_name in (
            "supporting_evidence",
            "opposing_evidence",
        ):
            values = item.get(field_name, [])
            if not isinstance(values, list):
                continue
            for evidence in values:
                valid = validate_quote_item(
                    evidence,
                    chunk_by_id,
                )
                if valid is not None:
                    claim[field_name].append(valid)

        defenses = item.get("defenses", [])
        if isinstance(defenses, list):
            for defense in defenses:
                if not isinstance(defense, dict):
                    continue
                effect = str(
                    defense.get(
                        "effect_candidate",
                        "UNCERTAIN",
                    )
                ).upper()
                if effect not in VALID_DEFENSE_EFFECTS:
                    effect = "UNCERTAIN"
                evidence_rows = []
                for evidence in defense.get(
                    "evidence",
                    [],
                ):
                    valid = validate_quote_item(
                        evidence,
                        chunk_by_id,
                    )
                    if valid is not None:
                        evidence_rows.append(valid)
                claim["defenses"].append({
                    "defense": truncate_words(
                        defense.get("defense", ""),
                        55,
                    ),
                    "effect_candidate": effect,
                    "evidence": evidence_rows,
                })

        graph["claims"].append(claim)

    graph["grounded_quote_count"] = sum(
        len(claim["supporting_evidence"])
        + len(claim["opposing_evidence"])
        + sum(
            len(defense["evidence"])
            for defense in claim["defenses"]
        )
        for claim in graph["claims"]
    ) + sum(
        len(graph[field_name])
        for field_name in (
            "appeal_scope",
            "prior_decisions",
            "prosecutor_opinions",
            "admissions",
            "objective_facts",
        )
    )

    if (
        require_a_claim
        and not any(
            claim["claim_type"] == "A_CLAIM"
            for claim in graph["claims"]
        )
    ):
        seed = build_query_seed_a_claim(
            case,
            {
                claim["claim_id"]
                for claim in graph["claims"]
            },
        )
        graph["claims"].insert(0, seed)
        graph["coverage_notes"].append(
            "Validator đã thêm A_CLAIM từ case_query vì model không xuất "
            "claim nguyên đơn theo schema."
        )
        graph["a_claim_seeded_from_query"] = True
    else:
        graph["a_claim_seeded_from_query"] = False

    return graph


def run_graph_extraction(
    cases: Sequence[CaseBundle],
    runtime: QwenRuntime,
    output_dir: Path,
    config: FinalOutcomeConfig,
) -> dict[str, dict[str, Any]]:
    output_dir.mkdir(parents=True, exist_ok=True)
    results = {}

    for position, case in enumerate(cases, start=1):
        path = output_dir / f"{case.case_id}.json"
        if path.exists():
            graph = read_json(path)
            case.graph = graph
            results[case.case_id] = graph
            continue

        user = f"""
CASE QUERY:
{case.case_query}

CLEAN QUERY:
{case.clean_query}

MANDATORY PLAINTIFF CLAIM SEED FROM CASE QUERY:
{extract_query_plaintiff_claim(case)[0]}

HINTS:
{build_hint_text(case.hints) or "(không có)"}

TOÀN BỘ UNIQUE CASE EVIDENCE:
{case.full_evidence_text}

Hãy tạo claim–evidence graph có exact quotes.
{graph_schema_text()}
"""
        print(
            f"[GRAPH {position}/{len(cases)}] "
            f"{case.case_id} chunks={len(case.chunks)}"
        )

        last_error = None
        graph = None
        for attempt in range(
            1,
            config.max_json_retries + 1,
        ):
            try:
                attempt_user = user
                if attempt > 1:
                    attempt_user += (
                        "\n\nREPAIR INSTRUCTION:\n"
                        "Output trước không qua validator. BẮT BUỘC có ít "
                        "nhất một claim với owner=\"A\" và "
                        "claim_type=\"A_CLAIM\", lấy nội dung yêu cầu "
                        "nguyên đơn từ CASE QUERY. Evidence arrays được phép "
                        "rỗng khi chưa có exact quote. Không đổi schema."
                    )

                raw = runtime.generate_json(
                    GRAPH_SYSTEM,
                    attempt_user,
                    max_new_tokens=(
                        config.extraction_max_new_tokens
                    ),
                    enable_thinking=(
                        config.extraction_enable_thinking
                    ),
                )
                graph = validate_graph(raw, case)
                break
            except Exception as exc:
                last_error = exc
                print(
                    f"  graph retry {attempt}: "
                    f"{type(exc).__name__}: {exc}"
                )

        if graph is None:
            raise RuntimeError(
                f"Graph extraction thất bại {case.case_id}: "
                f"{last_error}"
            )

        atomic_write_json(path, graph)
        case.graph = graph
        results[case.case_id] = graph

    return results


COVERAGE_SYSTEM = """
Bạn là bộ kiểm tra omission cho claim–evidence graph.

Đọc toàn bộ case evidence và graph hiện tại. Chỉ bổ sung claim hoặc evidence
quan trọng bị bỏ sót. Không dự đoán nhãn. Mọi bổ sung phải có chunk_id và quote
nguyên văn. Không lặp lại items đã có. Chỉ trả JSON theo cùng schema graph.
"""


def needs_coverage_audit(
    graph: dict[str, Any],
    config: FinalOutcomeConfig,
) -> bool:
    return (
        len(graph.get("claims", []))
        <= config.coverage_audit_min_claims
        or int(graph.get("grounded_quote_count", 0))
        < config.coverage_audit_min_grounded_quotes
        or graph.get("procedural_stage") == "APPEAL"
        and not graph.get("appeal_scope")
    )


def merge_graphs(
    base: dict[str, Any],
    addition: dict[str, Any],
) -> dict[str, Any]:
    merged = json.loads(json.dumps(base, ensure_ascii=False))

    existing_claim_text = {
        normalize_text(claim.get("claim", ""))
        for claim in merged.get("claims", [])
    }
    for claim in addition.get("claims", []):
        marker = normalize_text(claim.get("claim", ""))
        if marker and marker not in existing_claim_text:
            merged["claims"].append(claim)
            existing_claim_text.add(marker)

    for field_name in (
        "appeal_scope",
        "prior_decisions",
        "prosecutor_opinions",
        "admissions",
        "objective_facts",
        "coverage_notes",
    ):
        merged[field_name] = dedupe_preserve(
            merged.get(field_name, [])
            + addition.get(field_name, [])
        )

    merged["grounded_quote_count"] = sum(
        len(claim.get("supporting_evidence", []))
        + len(claim.get("opposing_evidence", []))
        + sum(
            len(defense.get("evidence", []))
            for defense in claim.get("defenses", [])
        )
        for claim in merged.get("claims", [])
    ) + sum(
        len(merged.get(field_name, []))
        for field_name in (
            "appeal_scope",
            "prior_decisions",
            "prosecutor_opinions",
            "admissions",
            "objective_facts",
        )
    )
    return merged


def run_conditional_coverage_audit(
    cases: Sequence[CaseBundle],
    runtime: QwenRuntime,
    output_dir: Path,
    config: FinalOutcomeConfig,
) -> None:
    if not config.run_conditional_coverage_audit:
        return

    output_dir.mkdir(parents=True, exist_ok=True)

    for case in cases:
        if not needs_coverage_audit(case.graph, config):
            continue

        path = output_dir / f"{case.case_id}.json"
        if path.exists():
            audited = read_json(path)
            case.graph = audited
            continue

        user = f"""
CASE QUERY:
{case.case_query}

GRAPH HIỆN TẠI:
{json.dumps(case.graph, ensure_ascii=False)}

TOÀN BỘ UNIQUE CASE EVIDENCE:
{case.full_evidence_text}

Chỉ trả các additions bị bỏ sót theo schema:
{graph_schema_text()}
"""
        print("[COVERAGE AUDIT]", case.case_id)

        last_error = None
        addition = None
        for attempt in range(
            1,
            config.max_json_retries + 1,
        ):
            try:
                raw = runtime.generate_json(
                    COVERAGE_SYSTEM,
                    user,
                    max_new_tokens=(
                        config.coverage_audit_max_new_tokens
                    ),
                    enable_thinking=False,
                )
                addition = validate_graph(
                    raw,
                    case,
                    require_a_claim=False,
                )
                break
            except Exception as exc:
                last_error = exc
                print(
                    f"  coverage retry {attempt}: "
                    f"{type(exc).__name__}: {exc}"
                )

        if addition is None:
            print(
                "  Bỏ qua coverage audit thất bại:",
                last_error,
            )
            continue

        case.graph = merge_graphs(
            case.graph,
            addition,
        )
        atomic_write_json(path, case.graph)


# =============================================================================
# Claim-specific law reranking
# =============================================================================

def law_document_text(article: LawArticle) -> str:
    return (
        f"law_id={article.law_id}; "
        f"{article.law_title}; Điều {article.article_number}; "
        f"aid={article.aid}\n{article.text}"
    )


def claim_query_text(
    case: CaseBundle,
    claim: dict[str, Any],
) -> str:
    support_quotes = " ".join(
        item.get("quote", "")
        for item in claim.get(
            "supporting_evidence",
            [],
        )[:6]
    )
    opposing_quotes = " ".join(
        item.get("quote", "")
        for item in claim.get(
            "opposing_evidence",
            [],
        )[:4]
    )
    defenses = " ".join(
        defense.get("defense", "")
        for defense in claim.get(
            "defenses",
            [],
        )[:4]
    )
    return (
        f"Vụ án: {case.clean_query}\n"
        f"Claim: {claim.get('claim')}\n"
        f"Remedy: {claim.get('remedy')}\n"
        f"Evidence ủng hộ: {support_quotes}\n"
        f"Evidence phản bác: {opposing_quotes}\n"
        f"Defenses: {defenses}"
    )


def run_claim_law_reranking(
    cases: Sequence[CaseBundle],
    article_by_aid: dict[int, LawArticle],
    runtime: QwenRerankerRuntime,
    output_dir: Path,
    config: FinalOutcomeConfig,
) -> dict[str, dict[str, list[dict[str, Any]]]]:
    output_dir.mkdir(parents=True, exist_ok=True)
    results = {}

    for position, case in enumerate(cases, start=1):
        path = output_dir / f"{case.case_id}.json"
        if path.exists():
            value = read_json(path)
            case.claim_laws = value
            results[case.case_id] = value
            continue

        law_candidates = [
            article_by_aid[aid]
            for aid in case.selected_law_aids
            if aid in article_by_aid
        ]

        claim_laws: dict[str, list[dict[str, Any]]] = {}
        print(
            f"[LAW RERANK {position}/{len(cases)}] "
            f"{case.case_id} claims={len(case.graph['claims'])} "
            f"laws={len(law_candidates)}"
        )

        for claim in case.graph.get("claims", []):
            claim_id = claim["claim_id"]
            if not law_candidates:
                claim_laws[claim_id] = []
                continue

            query = claim_query_text(case, claim)
            pairs = [
                (query, law_document_text(article))
                for article in law_candidates
            ]
            scores = runtime.score_pairs(pairs)
            ranked = sorted(
                zip(law_candidates, scores),
                key=lambda item: -item[1],
            )[:config.max_laws_per_claim]

            claim_laws[claim_id] = [
                {
                    "aid": article.aid,
                    "law_id": article.law_id,
                    "law_title": article.law_title,
                    "article_number": article.article_number,
                    "score": float(score),
                    "text": article.text,
                }
                for article, score in ranked
            ]

        atomic_write_json(path, claim_laws)
        case.claim_laws = claim_laws
        results[case.case_id] = claim_laws

    return results


# =============================================================================
# Three judge passes
# =============================================================================

JUDGE_SYSTEM_BASE = """
Bạn là thẩm phán dự đoán kết quả vụ án dân sự Việt Nam.

A luôn là NGUYÊN ĐƠN; B luôn là BỊ ĐƠN, kể cả ở phúc thẩm.

Bạn nhận:
- claim–evidence graph được grounded bằng exact quotes;
- toàn bộ raw case evidence;
- tối đa bốn điều luật được rerank riêng cho từng claim.

Quy tắc:
1. Không coi ý kiến Viện kiểm sát hoặc quyết định cấp dưới là kết quả cuối.
2. B_DEFENSE chỉ điều chỉnh A_CLAIM; không double-count như counterclaim.
3. B_COUNTERCLAIM chỉ là yêu cầu cứu trợ độc lập của B.
4. Đánh giá từng claim theo yếu tố pháp lý, burden of proof, defense và remedy.
5. Mỗi claim result phải dùng đúng claim_id trong graph.
6. Chỉ dùng chunk_id và aid có trong input.
7. Không dùng xác suất từ pipeline trước.
8. Chỉ trả JSON hợp lệ.
"""


PASS_INSTRUCTIONS = {
    "ELEMENTS_BURDEN": """
Góc phân tích: ELEMENTS VÀ BURDEN OF PROOF.
Với từng claim, xác định yếu tố pháp lý cần chứng minh, evidence nào đáp ứng,
evidence nào còn thiếu, defense có đánh bại hay chỉ giảm cứu trợ.
""",
    "ADVERSARIAL": """
Góc phân tích: ADVERSARIAL.
Xây strongest case for A và strongest case for B. Phân biệt allegation,
admission, objective evidence và contradiction. Kiểm tra nhầm vai A/B,
double-count defense và suy luận từ ý kiến Viện kiểm sát.
""",
    "REMEDY_OUTCOME": """
Góc phân tích: REMEDY VÀ KẾT QUẢ THỰC CHẤT.
Tập trung A thực tế có khả năng nhận tài sản/tiền/quyền gì, phần nào bị bác,
B có counterclaim được chấp nhận không, và bên nào thắng vấn đề có trọng lượng
lớn nhất. Không chỉ đếm số claim.
""",
}


def judge_schema_text() -> str:
    return """
Schema:
{
  "pass_name": "...",
  "claim_results": [
    {
      "claim_id": "...",
      "result": "GRANTED|PARTIALLY_GRANTED|REJECTED|UNCERTAIN",
      "relief_fraction": 0.0,
      "defense_effect": "DEFEAT_CLAIM|REDUCE_RELIEF|NO_EFFECT|UNCERTAIN",
      "case_evidence_ids": ["chunk_id"],
      "law_aids": [123],
      "short_reason": "tối đa 70 từ"
    }
  ],
  "direct_label": "A_WIN|PARTIAL_A_WIN|PARTIAL_B_WIN|B_WIN",
  "key_issue": "...",
  "confidence": "HIGH|MEDIUM|LOW"
}
relief_fraction phải nằm trong [0,1].
"""


def compact_graph_with_laws(case: CaseBundle) -> dict[str, Any]:
    graph = json.loads(
        json.dumps(case.graph, ensure_ascii=False)
    )
    for claim in graph.get("claims", []):
        claim["reranked_laws"] = case.claim_laws.get(
            claim["claim_id"],
            [],
        )
    return graph


def validate_judge_output(
    raw: dict[str, Any],
    case: CaseBundle,
    pass_name: str,
) -> dict[str, Any]:
    valid_claim_ids = {
        claim["claim_id"]
        for claim in case.graph.get("claims", [])
    }
    valid_chunk_ids = set(case.all_chunk_ids)
    valid_law_aids = set(case.selected_law_aids)

    rows = []
    seen = set()
    for item in raw.get("claim_results", []):
        if not isinstance(item, dict):
            continue
        claim_id = str(item.get("claim_id", "")).strip()
        if claim_id not in valid_claim_ids or claim_id in seen:
            continue
        seen.add(claim_id)

        result = str(
            item.get("result", "UNCERTAIN")
        ).upper()
        if result not in VALID_RESULTS:
            result = "UNCERTAIN"

        fraction = safe_float(
            item.get("relief_fraction"),
            RESULT_FRACTION[result],
        )
        fraction = clamp(fraction, 0.0, 1.0)

        defense_effect = str(
            item.get(
                "defense_effect",
                "UNCERTAIN",
            )
        ).upper()
        if defense_effect not in VALID_DEFENSE_EFFECTS:
            defense_effect = "UNCERTAIN"

        evidence_ids = [
            str(value)
            for value in item.get(
                "case_evidence_ids",
                [],
            )
            if str(value) in valid_chunk_ids
        ]
        law_aids = []
        for value in item.get("law_aids", []):
            try:
                aid = int(value)
            except (TypeError, ValueError):
                continue
            if aid in valid_law_aids:
                law_aids.append(aid)

        rows.append({
            "claim_id": claim_id,
            "result": result,
            "relief_fraction": fraction,
            "defense_effect": defense_effect,
            "case_evidence_ids": dedupe_preserve(
                evidence_ids
            ),
            "law_aids": dedupe_preserve(law_aids),
            "short_reason": truncate_words(
                item.get("short_reason", ""),
                75,
            ),
        })

    direct_label = str(
        raw.get("direct_label", "")
    ).upper()
    if direct_label not in LABELS:
        raise ValueError("direct_label không hợp lệ.")

    if not rows:
        raise ValueError("Không có claim_results hợp lệ.")

    return {
        "case_id": case.case_id,
        "pass_name": pass_name,
        "claim_results": rows,
        "direct_label": direct_label,
        "key_issue": truncate_words(
            raw.get("key_issue", ""),
            90,
        ),
        "confidence": str(
            raw.get("confidence", "LOW")
        ).upper(),
    }


def run_judge_passes(
    cases: Sequence[CaseBundle],
    runtime: QwenRuntime,
    output_dir: Path,
    config: FinalOutcomeConfig,
) -> dict[str, list[dict[str, Any]]]:
    output_dir.mkdir(parents=True, exist_ok=True)
    results: dict[str, list[dict[str, Any]]] = {}

    for case_position, case in enumerate(cases, start=1):
        case_results = []

        for pass_name in config.judge_passes:
            pass_dir = output_dir / pass_name.lower()
            pass_dir.mkdir(parents=True, exist_ok=True)
            path = pass_dir / f"{case.case_id}.json"

            if path.exists():
                case_results.append(read_json(path))
                continue

            user = f"""
PASS:
{pass_name}

HƯỚNG DẪN RIÊNG:
{PASS_INSTRUCTIONS[pass_name]}

CASE QUERY:
{case.case_query}

CLAIM–EVIDENCE GRAPH + CLAIM-SPECIFIC LAWS:
{json.dumps(compact_graph_with_laws(case), ensure_ascii=False)}

TOÀN BỘ UNIQUE CASE EVIDENCE:
{case.full_evidence_text}

Hãy đánh giá từng claim và trả JSON.
{judge_schema_text()}
"""
            print(
                f"[JUDGE {case_position}/{len(cases)}] "
                f"{case.case_id} {pass_name}"
            )

            last_error = None
            parsed = None
            for attempt in range(
                1,
                config.max_json_retries + 1,
            ):
                try:
                    raw = runtime.generate_json(
                        JUDGE_SYSTEM_BASE,
                        user,
                        max_new_tokens=(
                            config.judge_max_new_tokens
                        ),
                        enable_thinking=(
                            config.judge_enable_thinking
                        ),
                    )
                    parsed = validate_judge_output(
                        raw,
                        case,
                        pass_name,
                    )
                    break
                except Exception as exc:
                    last_error = exc
                    print(
                        f"  judge retry {attempt}: "
                        f"{type(exc).__name__}: {exc}"
                    )

            if parsed is None:
                raise RuntimeError(
                    f"Judge thất bại {case.case_id} "
                    f"{pass_name}: {last_error}"
                )

            atomic_write_json(path, parsed)
            case_results.append(parsed)

        results[case.case_id] = case_results

    return results


# =============================================================================
# Deterministic claim-to-label mapping and consensus
# =============================================================================

def claim_by_id(case: CaseBundle) -> dict[str, dict[str, Any]]:
    return {
        claim["claim_id"]: claim
        for claim in case.graph.get("claims", [])
    }


def rule_probabilities(
    case: CaseBundle,
    judge_result: dict[str, Any],
    config: FinalOutcomeConfig,
) -> dict[str, float]:
    claims = claim_by_id(case)
    result_by_id = {
        row["claim_id"]: row
        for row in judge_result.get(
            "claim_results",
            [],
        )
    }

    a_points = 0.0
    b_points = 0.0
    a_total = 0.0
    b_total = 0.0
    a_core_fractions = []
    b_core_fractions = []

    for claim_id, claim in claims.items():
        row = result_by_id.get(claim_id)
        if row is None:
            fraction = 0.5
        else:
            fraction = clamp(
                safe_float(
                    row.get("relief_fraction"),
                    RESULT_FRACTION.get(
                        row.get("result"),
                        0.5,
                    ),
                ),
                0.0,
                1.0,
            )

        weight = IMPORTANCE_WEIGHT.get(
            claim.get("importance", "SECONDARY"),
            1.0,
        )

        if claim.get("claim_type") == "A_CLAIM":
            a_points += weight * fraction
            b_points += weight * (1.0 - fraction)
            a_total += weight
            if claim.get("importance") == "CORE":
                a_core_fractions.append(fraction)

        elif claim.get("claim_type") == "B_COUNTERCLAIM":
            b_points += weight * fraction
            a_points += weight * (1.0 - fraction)
            b_total += weight
            if claim.get("importance") == "CORE":
                b_core_fractions.append(fraction)

    total = a_points + b_points
    if total <= 0:
        return {label: 0.25 for label in LABELS}

    a_share = a_points / total
    a_core = (
        float(np.mean(a_core_fractions))
        if a_core_fractions
        else a_share
    )
    b_core = (
        float(np.mean(b_core_fractions))
        if b_core_fractions
        else 1.0 - a_share
    )

    scores = {
        "A_WIN": max(
            0.0,
            1.0 - abs(a_share - 0.94) / 0.40,
        ),
        "PARTIAL_A_WIN": max(
            0.0,
            1.0 - abs(a_share - 0.67) / 0.30,
        ),
        "PARTIAL_B_WIN": max(
            0.0,
            1.0 - abs(a_share - 0.35) / 0.28,
        ),
        "B_WIN": max(
            0.0,
            1.0 - abs(a_share - 0.06) / 0.38,
        ),
    }

    if (
        a_core >= config.a_win_core_threshold
        and b_core <= 0.25
    ):
        scores["A_WIN"] += 0.55
    elif a_core >= 0.50:
        scores["PARTIAL_A_WIN"] += 0.28

    if (
        a_core <= config.b_win_core_threshold
        and b_core >= 0.75
    ):
        scores["B_WIN"] += 0.55
    elif a_core < 0.50:
        scores["PARTIAL_B_WIN"] += 0.28

    values = np.asarray(
        [max(1e-6, scores[label]) for label in LABELS],
        dtype=np.float64,
    )
    values /= values.sum()
    return {
        label: float(values[index])
        for index, label in enumerate(LABELS)
    }


def rule_label(
    case: CaseBundle,
    judge_result: dict[str, Any],
    config: FinalOutcomeConfig,
) -> str:
    return top_label(
        rule_probabilities(
            case,
            judge_result,
            config,
        )
    )


def build_votes(
    case: CaseBundle,
    judge_results: Sequence[dict[str, Any]],
    config: FinalOutcomeConfig,
) -> tuple[Counter, list[dict[str, Any]]]:
    votes: Counter = Counter()
    details = []

    for result in judge_results:
        direct = (result.get("direct_label") or result.get("prediction") or result.get("label") or result.get("verdict_label") or "PARTIAL_A_WIN")
        rule = rule_label(case, result, config)

        # Rule label carries more weight because it is derived from the
        # per-claim matrix and does not use self-reported probabilities.
        votes[rule] += 2
        votes[direct] += 1

        details.append({
            "pass_name": result.get("pass_name", "judge_pass"),
            "direct_label": direct,
            "rule_label": rule,
            "rule_probabilities": rule_probabilities(
                case,
                result,
                config,
            ),
        })

    return votes, details


def consensus_candidates(
    votes: Counter,
) -> tuple[list[str], int]:
    ordered = sorted(
        LABELS,
        key=lambda label: (
            -votes[label],
            LABEL_INDEX[label],
        ),
    )
    margin = votes[ordered[0]] - votes[ordered[1]]
    return ordered[:2], int(margin)


# =============================================================================
# Independent verifier for ambiguous cases
# =============================================================================

VERIFIER_SYSTEM = """
Bạn là verifier độc lập cho dự đoán kết quả vụ án dân sự Việt Nam.

Chỉ chọn giữa hai nhãn ứng viên. Không viết lại toàn bộ vụ án.

Kiểm tra:
1. A là nguyên đơn, B là bị đơn.
2. Claim cốt lõi nào của A được chấp nhận và ở mức nào.
3. B_DEFENSE có bị double-count như counterclaim không.
4. B có counterclaim cứu trợ độc lập hay chỉ defense.
5. Có nhầm ý kiến Viện kiểm sát/quyết định cấp dưới thành kết quả cuối không.
6. Nhãn có nhất quán với claim-result matrices không.
7. Counterfactual: nếu defense chính của B thành công hoặc thất bại thì nhãn
   partial nghiêng về bên nào.
8. Role-bias check: không ưu tiên A chỉ vì A là nguyên đơn.
Chỉ trả JSON.
"""


def compact_verifier_packet(
    case: CaseBundle,
    judge_results: Sequence[dict[str, Any]],
    vote_details: Sequence[dict[str, Any]],
    candidate_labels: Sequence[str],
) -> str:
    graph = json.loads(
        json.dumps(case.graph, ensure_ascii=False)
    )

    # Aya has an 8K context. Keep all claims and grounded quotes, omit the full
    # raw evidence because exact quotes are already in the graph.
    for claim in graph.get("claims", []):
        claim["supporting_evidence"] = claim.get(
            "supporting_evidence",
            [],
        )[:6]
        claim["opposing_evidence"] = claim.get(
            "opposing_evidence",
            [],
        )[:4]
        for defense in claim.get("defenses", []):
            defense["evidence"] = defense.get(
                "evidence",
                [],
            )[:3]
        claim["reranked_laws"] = case.claim_laws.get(
            claim["claim_id"],
            [],
        )[:3]

    return (
        f"HAI NHÃN ỨNG VIÊN:\n"
        f"1. {candidate_labels[0]} — "
        f"{LABEL_DEFINITIONS[candidate_labels[0]]}\n"
        f"2. {candidate_labels[1]} — "
        f"{LABEL_DEFINITIONS[candidate_labels[1]]}\n\n"
        f"GRAPH:\n{json.dumps(graph, ensure_ascii=False)}\n\n"
        f"JUDGE RESULTS:\n"
        f"{json.dumps(list(judge_results), ensure_ascii=False)}\n\n"
        f"VOTE DETAILS:\n"
        f"{json.dumps(list(vote_details), ensure_ascii=False)}\n\n"
        "Trả JSON:\n"
        "{\n"
        f'  "prediction": "{candidate_labels[0]}|{candidate_labels[1]}",\n'
        '  "winning_claim_ids": ["..."],\n'
        '  "failed_claim_ids": ["..."],\n'
        '  "reason": "tối đa 120 từ"\n'
        "}"
    )


def run_verifier(
    cases: Sequence[CaseBundle],
    all_judge_results: dict[str, list[dict[str, Any]]],
    runtime: AyaVerifierRuntime,
    output_dir: Path,
    config: FinalOutcomeConfig,
) -> dict[str, dict[str, Any]]:
    output_dir.mkdir(parents=True, exist_ok=True)
    results = {}

    for case in cases:
        judge_results = all_judge_results[case.case_id]
        votes, details = build_votes(
            case,
            judge_results,
            config,
        )
        candidates, margin = consensus_candidates(votes)

        rule_labels = {
            row["rule_label"]
            for row in details
        }
        direct_labels = {
            row["direct_label"]
            for row in details
        }
        needs_verifier = (
            margin < config.verifier_vote_margin_threshold
            or (
                config.verifier_rule_disagreement
                and (
                    len(rule_labels) > 1
                    or len(direct_labels) > 1
                )
            )
        )

        if not needs_verifier:
            continue

        path = output_dir / f"{case.case_id}.json"
        if path.exists():
            results[case.case_id] = read_json(path)
            continue

        user = compact_verifier_packet(
            case,
            judge_results,
            details,
            candidates,
        )
        print(
            "[VERIFIER]",
            case.case_id,
            candidates,
            "vote_margin=",
            margin,
        )

        last_error = None
        parsed = None
        for attempt in range(
            1,
            config.max_json_retries + 1,
        ):
            try:
                raw = runtime.generate_json(
                    VERIFIER_SYSTEM,
                    user,
                    config.verifier_max_new_tokens,
                )
                prediction = str(
                    raw.get("prediction", "")
                ).upper()
                if prediction not in candidates:
                    raise ValueError(
                        "Verifier chọn nhãn ngoài hai candidates."
                    )
                parsed = {
                    "case_id": case.case_id,
                    "prediction": prediction,
                    "candidate_labels": candidates,
                    "vote_margin": margin,
                    "winning_claim_ids": [
                        str(value)
                        for value in raw.get(
                            "winning_claim_ids",
                            [],
                        )
                    ],
                    "failed_claim_ids": [
                        str(value)
                        for value in raw.get(
                            "failed_claim_ids",
                            [],
                        )
                    ],
                    "reason": truncate_words(
                        raw.get("reason", ""),
                        130,
                    ),
                    "verifier_model": runtime.model_id,
                }
                break
            except Exception as exc:
                last_error = exc
                print(
                    f"  verifier retry {attempt}: "
                    f"{type(exc).__name__}: {exc}"
                )

        if parsed is None:
            raise RuntimeError(
                f"Verifier thất bại {case.case_id}: "
                f"{last_error}"
            )

        atomic_write_json(path, parsed)
        results[case.case_id] = parsed

    return results


# =============================================================================
# Submission and diagnostics
# =============================================================================

def build_law_evidence_objects(
    aids: Sequence[int],
    article_by_aid: dict[int, LawArticle],
) -> list[dict[str, Any]]:
    output = []
    seen = set()

    for value in aids:
        aid = int(value)
        article = article_by_aid.get(aid)
        if article is None:
            raise ValueError(
                f"aid {aid} không tồn tại trong corpus."
            )
        key = (article.law_id, aid)
        if key in seen:
            continue
        seen.add(key)
        output.append({
            "law_id": article.law_id,
            "aid": aid,
        })

    return output


def final_prediction_for_case(
    case: CaseBundle,
    judge_results: Sequence[dict[str, Any]],
    verifier: Optional[dict[str, Any]],
    config: FinalOutcomeConfig,
) -> tuple[str, dict[str, Any]]:
    votes, details = build_votes(
        case,
        judge_results,
        config,
    )
    candidates, margin = consensus_candidates(votes)

    prediction = (
        verifier["prediction"]
        if verifier is not None
        else candidates[0]
    )

    return prediction, {
        "votes": dict(votes),
        "vote_details": details,
        "candidate_labels": candidates,
        "vote_margin": margin,
        "verifier_used": verifier is not None,
        "verifier": verifier,
    }


def validate_submission(
    rows: Sequence[dict[str, Any]],
    originals: Sequence[dict[str, Any]],
    article_by_aid: dict[int, LawArticle],
) -> dict[str, Any]:
    errors = []
    expected_ids = [
        str(item["case_id"])
        for item in originals
    ]
    actual_ids = [
        str(row.get("case_id"))
        for row in rows
    ]

    if actual_ids != expected_ids:
        errors.append(
            "Tập hoặc thứ tự case_id không khớp private input."
        )
    if len(actual_ids) != len(set(actual_ids)):
        errors.append("Trùng case_id.")

    label_counts = Counter()
    case_counts = []
    law_counts = []

    for row in rows:
        case_id = row.get("case_id")
        prediction = row.get("prediction")
        if prediction not in LABELS:
            errors.append(
                f"{case_id}: prediction không hợp lệ."
            )
        else:
            label_counts[prediction] += 1

        case_evidence = row.get("case_evidence")
        if not isinstance(case_evidence, list):
            errors.append(
                f"{case_id}: case_evidence không phải list."
            )
            case_evidence = []
        if len(case_evidence) != len(set(case_evidence)):
            errors.append(
                f"{case_id}: trùng case_evidence."
            )
        case_counts.append(len(case_evidence))

        law_evidence = row.get("law_evidence")
        if not isinstance(law_evidence, list):
            errors.append(
                f"{case_id}: law_evidence không phải list."
            )
            law_evidence = []

        pairs = []
        for item in law_evidence:
            if not isinstance(item, dict):
                errors.append(
                    f"{case_id}: law item không phải object."
                )
                continue
            if set(item) != {"law_id", "aid"}:
                errors.append(
                    f"{case_id}: law item phải có đúng law_id và aid."
                )
            try:
                aid = int(item.get("aid"))
            except (TypeError, ValueError):
                errors.append(
                    f"{case_id}: aid không phải integer."
                )
                continue
            article = article_by_aid.get(aid)
            if article is None:
                errors.append(
                    f"{case_id}: aid {aid} không có trong corpus."
                )
                continue
            if item.get("law_id") != article.law_id:
                errors.append(
                    f"{case_id}: law_id không khớp aid {aid}."
                )
            pairs.append((item.get("law_id"), aid))

        if len(pairs) != len(set(pairs)):
            errors.append(
                f"{case_id}: trùng law evidence."
            )
        law_counts.append(len(law_evidence))

    return {
        "valid": not errors,
        "errors": errors,
        "case_count": len(rows),
        "label_distribution": dict(label_counts),
        "mean_case_evidence": (
            float(np.mean(case_counts))
            if case_counts else 0.0
        ),
        "min_case_evidence": min(case_counts) if case_counts else 0,
        "max_case_evidence": max(case_counts) if case_counts else 0,
        "mean_law_evidence": (
            float(np.mean(law_counts))
            if law_counts else 0.0
        ),
    }


def build_submission(
    cases: Sequence[CaseBundle],
    originals: Sequence[dict[str, Any]],
    article_by_aid: dict[int, LawArticle],
    judge_results: dict[str, list[dict[str, Any]]],
    verifier_results: dict[str, dict[str, Any]],
    output_dir: Path,
    config: FinalOutcomeConfig,
) -> tuple[list[dict[str, Any]], list[dict[str, Any]]]:
    output_dir.mkdir(parents=True, exist_ok=True)
    case_by_id = {
        case.case_id: case
        for case in cases
    }

    rows = []
    diagnostics = []

    for original in originals:
        case_id = str(original["case_id"])
        case = case_by_id[case_id]
        prediction, diagnostic = final_prediction_for_case(
            case,
            judge_results[case_id],
            verifier_results.get(case_id),
            config,
        )

        # User requested all retrieved unique case evidence. This is also the
        # evidence set that was used by the graph extraction stage.
        case_evidence = dedupe_preserve(
            case.all_chunk_ids
        )

        row = {
            "case_id": case_id,
            "prediction": prediction,
            "case_evidence": case_evidence,
            "law_evidence": build_law_evidence_objects(
                case.selected_law_aids,
                article_by_aid,
            ),
        }
        rows.append(row)

        diagnostics.append({
            "case_id": case_id,
            "prediction": prediction,
            "claim_count": len(
                case.graph.get("claims", [])
            ),
            "grounded_quote_count": case.graph.get(
                "grounded_quote_count",
                0,
            ),
            "case_evidence_count": len(case_evidence),
            "law_evidence_count": len(
                row["law_evidence"]
            ),
            **diagnostic,
        })

    if (
        config.expected_private_cases
        and len(rows) != config.expected_private_cases
    ):
        raise AssertionError(
            f"Expected {config.expected_private_cases} cases, "
            f"nhận {len(rows)}."
        )

    atomic_write_json(
        output_dir / "submission.json",
        rows,
    )
    atomic_write_json(
        output_dir / "submission_diagnostics.json",
        diagnostics,
    )

    validation = validate_submission(
        rows,
        originals,
        article_by_aid,
    )
    atomic_write_json(
        output_dir / "submission_validation.json",
        validation,
    )
    if not validation["valid"]:
        raise RuntimeError(
            "Submission không hợp lệ: "
            + "; ".join(validation["errors"])
        )

    return rows, diagnostics


# =============================================================================
# Full pipeline
# =============================================================================

def run_final_pipeline(
    *,
    private_test_path: Path,
    law_corpus_path: Path,
    case_retrieval_dir: Path,
    private_case_and_law_path: Path,
    work_dir: Path,
    config: FinalOutcomeConfig,
) -> dict[str, Any]:
    work_dir.mkdir(parents=True, exist_ok=True)

    safe_config = asdict(config)
    safe_config["hf_token"] = None
    atomic_write_json(
        work_dir / "config.json",
        safe_config,
    )

    article_by_aid = load_law_corpus(
        law_corpus_path
    )
    cases, originals = load_private_cases(
        private_test_path,
        case_retrieval_dir,
        private_case_and_law_path,
    )

    for case in cases:
        build_full_evidence_text(case)

    atomic_write_json(
        work_dir / "case_manifest.json",
        [
            {
                "case_id": case.case_id,
                "unique_chunk_count": len(
                    case.all_chunk_ids
                ),
                "selected_law_count": len(
                    case.selected_law_aids
                ),
                "evidence_words": len(
                    case.full_evidence_text.split()
                ),
            }
            for case in cases
        ],
    )

    # ------------------------------------------------------------------
    # Stage 1: full-evidence grounded graph
    # ------------------------------------------------------------------
    qwen = QwenRuntime(config)
    qwen.load()

    run_graph_extraction(
        cases,
        qwen,
        work_dir / "graphs",
        config,
    )
    run_conditional_coverage_audit(
        cases,
        qwen,
        work_dir / "graphs_audited",
        config,
    )

    # Save current graph state even when no audit was required.
    for case in cases:
        atomic_write_json(
            work_dir
            / "graphs_final"
            / f"{case.case_id}.json",
            case.graph,
        )

    qwen.unload()

    # ------------------------------------------------------------------
    # Stage 2: claim-specific law reranking
    # ------------------------------------------------------------------
    reranker = QwenRerankerRuntime(config)
    reranker.load()

    run_claim_law_reranking(
        cases,
        article_by_aid,
        reranker,
        work_dir / "claim_laws",
        config,
    )

    reranker.unload()

    # ------------------------------------------------------------------
    # Stage 3: three complementary Qwen judge passes
    # ------------------------------------------------------------------
    qwen = QwenRuntime(config)
    qwen.load()

    judge_results = run_judge_passes(
        cases,
        qwen,
        work_dir / "judge_passes",
        config,
    )

    qwen.unload()

    # ------------------------------------------------------------------
    # Stage 4: independent verifier for ambiguous cases
    # ------------------------------------------------------------------
    verifier = AyaVerifierRuntime(config)
    verifier.load()

    verifier_results = run_verifier(
        cases,
        judge_results,
        verifier,
        work_dir / "verifier",
        config,
    )
    verifier_model_used = verifier.model_id
    verifier_fallback_used = verifier.using_fallback
    verifier.unload()

    # ------------------------------------------------------------------
    # Stage 5: deterministic final output
    # ------------------------------------------------------------------
    rows, diagnostics = build_submission(
        cases,
        originals,
        article_by_aid,
        judge_results,
        verifier_results,
        work_dir / "final",
        config,
    )

    summary = {
        "public_data_used": False,
        "private_case_count": len(cases),
        "qwen_model": config.qwen_model_id,
        "reranker_model": config.reranker_model_id,
        "verifier_model_used": verifier_model_used,
        "verifier_fallback_used": verifier_fallback_used,
        "verifier_case_count": len(verifier_results),
        "prediction_distribution": dict(
            Counter(
                row["prediction"]
                for row in rows
            )
        ),
        "mean_claim_count": float(
            np.mean([
                row["claim_count"]
                for row in diagnostics
            ])
        ),
        "mean_grounded_quotes": float(
            np.mean([
                row["grounded_quote_count"]
                for row in diagnostics
            ])
        ),
        "mean_case_evidence": float(
            np.mean([
                row["case_evidence_count"]
                for row in diagnostics
            ])
        ),
        "submission_path": str(
            work_dir / "final" / "submission.json"
        ),
    }
    atomic_write_json(
        work_dir / "run_summary.json",
        summary,
    )
    return summary
````
